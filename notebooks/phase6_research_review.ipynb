### Quick help: choosing a decision

1. **Accept:** The label is **Intended Hadith quotation**, the writer clearly quotes the highlighted words, and the boundaries are right. Accept also applies when the label is **Not intended** and you agree.
2. **Correct:** A supposed quotation is actually ordinary prose. Choose Correct, then **Not intended**. Keep this useful negative example rather than rejecting it. You can also correct recoverable boundaries; the optional boundary editor shows a preview.
3. **Reject / Uncertain:** Reject an unusable fragment that cannot be meaningfully classified. Choose Uncertain when the words are meaningful but you cannot tell whether they are intended as a quotation. Add a short note if possible.

**Whole-paragraph review** asks whether the entire paragraph contains a Hadith quotation. **Highlighted-passage review** asks only about the marked words, in context. Matching words in a source does not by itself prove quotation intent or authenticate a Hadith.

**Save & Next** saves and moves on; **Save Progress** saves without moving. **Skip** leaves the saved decision alone and discards unsaved edits. **Previous** revisits items. Use **Find or filter review items** to revisit saved decisions. Uncertain counts as reviewed but remains unresolved.

Sources, configurations, uncertainty flags, offsets, alignment details and technical IDs are available under **Advanced Details**. Add a missed quotation through the separate collapsed panel. Offset edits use original Python characters, with exclusive ends, and preserve Arabic marks and whitespace.
