### Two simple questions

- **Textual correspondence: Yes / No / Uncertain.** Does the input agree with the displayed source passage, allowing the stated character edits and missing source words? Arabic diacritics and equivalent letter forms can differ without using an edit.
- **Boundaries: Yes / Correct / Uncertain.** Do both passages start and end appropriately? Correct opens an optional offset editor with previews. Its end offsets are exclusive Python character positions in the original text, including diacritics.

The source is never rewritten. Pink source words (or symbols containing them) are omitted from the input. Each pair states both the allowed changes and the changes used by the displayed correspondence. Other sources and every supporting occurrence remain available under Advanced evidence; you do not need to review them individually.

**Save & Next** advances once the questions on this input are answered. For an input with no detections, it only acknowledges the screen. **Skip for now** saves valid partial decisions and moves on. Use **Previous** or **All inputs** to revisit work. Uncertain is saved but unresolved. A boundary correction is not complete until it contains valid changed offsets; unfinished correction choices are saved as drafts. Notes save when you leave the notes field or press a navigation button.
